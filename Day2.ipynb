```python
#4. Write a program to print the first 15 terms of the Fibonacci series. The sequence starts with 0 and 1 and each subsequent number is the sum of the two proceeding ones.
n=15
a=0
b=1
for i in range(n):
    print(a,end=" ")
    c=a+b
    a=b
    b=c
```

```python
#5. Write a program that takes a year as input and determines if it is a leap year.
year=int(input("Enter a year:"))
print(year)
if year%400==0:
    print("Leap Year")
elif year%4==0 and year%100!=0:
    print("Leap Year")
else:
    print("Not a leap year")   
         
```

```python
#7. Take two lists and find the elements that appear in both. Use sets to perform the operation.
int_a=[1,2,3,4,5]
int_b=[4,5,6,7,8]
a=set(int_a)
b=set(int_b)
c=a.intersection(b)
d=list(c)
print(d)
```

```python
#8. Start with a list of 10 numbers. Iterate through them and sort them into two separate lists: one for even numbers and one for odd numbers.
numbers=[12,7,34,21,5,10,8,3,19,2]
even=[]
odd=[]
for item in numbers:
    if item%2==0:
        even.append(item)
    else:
        odd.append(item)    
print(f"even:{even},odd:{odd}")        
```
