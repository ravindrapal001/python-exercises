```python
#16. Write a program to check if a given number is a palindrome (reads the same forwards and backwards).
number=int(input("Enter a number:"))
original=number
reverse=0
while number>0:
    digit=number%10
    reverse=reverse*10+digit
    number=number//10   
print(reverse)     
if original==reverse:
    print("Palindrome")
else: 
    print("Not Palindrome")    
```
